```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import warnings
import plotly.express as px

warnings.filterwarnings('ignore')


df = pd.read_csv("social.csv")

# Display top records
print(df.head())

# Selecting relevant features
X = df[['Age', 'EstimatedSalary']]

# Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Using Elbow Method to find optimal clusters
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot Elbow Graph
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()

# Applying KMeans with optimal clusters (e.g., 4)
kmeans = KMeans(n_clusters=4, init='k-means++', random_state=42)
y_kmeans = kmeans.fit_predict(X_scaled)

# Visualizing the clusters
plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=y_kmeans, palette='Set1')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            s=300, c='yellow', label='Centroids', edgecolor='black')
plt.title('User Segments based on Age and Salary')
plt.xlabel('Age (scaled)')
plt.ylabel('Estimated Salary (scaled)')
plt.legend()
plt.show()

# Add Gender to visualization (if 'Gender' column exists)
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
    sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=y_kmeans, style=df['Gender'], palette='Set1')

# PCA for dimensionality reduction
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df['Cluster'] = y_kmeans
print(df.groupby('Cluster').mean())

# Interactive Visualization with Plotly
fig = px.scatter(x=X_scaled[:,0], y=X_scaled[:,1], color=y_kmeans.astype(str), 
                 title='Interactive User Segments')
fig.show()

# For predicting new data
def predict_cluster(new_data):
    # Scale the new data using the previously fitted scaler
    new_data_scaled = scaler.transform([new_data])
    
    # Predict the cluster using the trained KMeans model
    cluster = kmeans.predict(new_data_scaled)
    return cluster

# Example: Predicting the cluster for a new user
new_user = [25, 50000]  # Example input: [Age, Estimated Salary]
predicted_cluster = predict_cluster(new_user)
print(f"The new user belongs to Cluster {predicted_cluster[0]}")

# Visualizing the new data point along with existing clusters
plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=y_kmeans, palette='Set1')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            s=300, c='yellow', label='Centroids', edgecolor='black')

# Plot the new user
new_user_scaled = scaler.transform([new_user])
plt.scatter(new_user_scaled[0, 0], new_user_scaled[0, 1], c='red', s=200, label='New User', marker='X')

plt.title('User Segments with New User Prediction')
plt.xlabel('Age (scaled)')
plt.ylabel('Estimated Salary (scaled)')
plt.legend()
plt.show()
```
